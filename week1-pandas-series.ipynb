- a series is basically a fancy list/array where theres a customizeable index

```python
# Load packages
import pandas as pd
import numpy as np
```

```python
s = pd.Series([1, 2, 999, 999], index = ['A', 'B', 'C', 'D'])
```

check-in exercise 1:

```python
# Replace all values greater than equal to 999 with NaN
s.mask(s >= 999)
```

```python
# Initialize dictionary with columns' data 
d = {'col_name_1' : pd.Series(np.arange(3)),
     'col_name_2' : pd.Series([3.1, 3.2, 3.3]),
     }

# Create data frame
df = pd.DataFrame(d)
df
```

```python
# Change index
df.index = ['a','b','c']
df
```

check in exercise 2:

```python
# Change col names
df.columns = ['c1', 'c2']
df
```

### Summary

- series are like the building blocks or atoms of pandas
- a df is a library of series
- series are lists but u can customize the index
- series methods are most often applied to a column of a df
